```python
import pandas as pd
import numpy as np
from pathlib import Path
from PIL import Image
from IPython.display import display
# glob help with loading images in both flip and not flip using */*
train_files = list(Path("../data/training").glob("*/*"))
test_files = list(Path("../data/training").glob("*/*"))

train_images = [Image.open(file) for file in train_files]
test_images = [Image.open(file) for file in test_files]
train_labels = [file.parent.name for file in train_files]
test_labels = [file.parent.name for file in test_files]
```

```python
results = [
    {
        "file": file.name,
        "label": file.parent.name,
        "width": img.width,
        "height": img.height,
        "aspect_ratio": img.width / img.height
    }
    for file in train_files
    for img in [Image.open(file)]
]

image_info = pd.DataFrame(results)
```
